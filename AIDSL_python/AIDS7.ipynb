```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ----------------------------------
# PART A: NUMPY ARRAYS
# ----------------------------------

# Create arrays
arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)

# Basic operations
print("Addition:", arr + 5)
print("Multiplication:", arr * 2)
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))

# Slicing
print("First three elements:", arr[:3])

# Reshape
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Matrix:\n", matrix)
print("Transpose:\n", matrix.T)

# Visualize array
plt.figure(figsize=(7, 4))
plt.plot(arr, marker='o')
plt.title("NumPy Array Visualization")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.show()


# ----------------------------------
# PART B: PANDAS DATAFRAME
# ----------------------------------

df = pd.read_csv("StudentsPerformance.csv")

print("\nFirst five rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

print("\nStatistics:")
print(df.describe())

# Select columns
student_data = df[["gender", "math score", "reading score"]]

print(student_data.head())


# ----------------------------------
# PART C: HANDLE MISSING VALUES
# ----------------------------------

# Create a missing value for demonstration
student_data.loc[0, "math score"] = np.nan

print("\nMissing values:")
print(student_data.isnull().sum())

# Replace missing numerical values with mean
student_data["math score"] = student_data["math score"].fillna(
    student_data["math score"].mean()
)

print("\nAfter handling missing values:")
print(student_data.isnull().sum())


# ----------------------------------
# PART D: MERGE DATAFRAMES
# ----------------------------------

# Create two related DataFrames
df1 = df[["gender", "math score"]]
df2 = df[["reading score", "writing score"]]

# Merge using their common row index
merged_df = pd.merge(
    df1,
    df2,
    left_index=True,
    right_index=True
)

print("\nMerged DataFrame:")
print(merged_df.head())


# ----------------------------------
# PART E: VISUALIZATION
# ----------------------------------

sns.barplot(data=df, x="gender", y="math score")

plt.title("Average Math Scores by Gender")
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------- NUMPY ----------------

# Create a NumPy array
arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)

# Basic operations
print("Addition:", arr + 5)
print("Multiplication:", arr * 2)
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("First three elements:", arr[:3])

# Create a 2D array
matrix = np.array([[1, 2, 3], [4, 5, 6]])

print("\nMatrix:\n", matrix)
print("Transpose:\n", matrix.T)

# Visualize NumPy array
plt.plot(arr, marker='o')
plt.title("NumPy Array Visualization")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.show()


# ---------------- PANDAS ----------------

# Create DataFrame using dictionary
data = {
    "Name": ["Amit", "Riya", "Rahul", "Neha", "Karan"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 75, 95, 80],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
}

df = pd.DataFrame(data)

# Display and explore DataFrame
print("\nDataFrame:")
print(df)

print("\nFirst five rows:")
print(df.head())

print("\nDataFrame Information:")
print(df.info())

print("\nStatistical Description:")
print(df.describe())

# Select columns
print("\nNames:")
print(df["Name"])

# Filter rows
print("\nStudents with marks above 80:")
print(df[df["Marks"] > 80])

# Add a new column
df["Result"] = ["Pass", "Pass", "Pass", "Pass", "Pass"]

# Handle missing values
df.loc[2, "Marks"] = np.nan

print("\nMissing Values:")
print(df.isnull().sum())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("\nAfter Handling Missing Values:")
print(df)

# Create two DataFrames and merge them
df1 = df[["Name", "Age"]]
df2 = df[["Name", "Marks"]]

merged_df = pd.merge(df1, df2, on="Name")

print("\nMerged DataFrame:")
print(merged_df)

# Visualize using Seaborn
sns.barplot(data=df, x="Name", y="Marks")
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```
