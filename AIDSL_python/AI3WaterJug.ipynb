```python
import math
from collections import deque
```

```python
def water_jug(x, y, g): 
    if g > max(x, y) or g % math.gcd(x, y) != 0:
        return None

    parent = {
        (0, 0): None
    }
    queue = deque([(0,0)])

    while queue: 
        current = queue.popleft()
        if goal_test(current, g): 
            return reconstruct_path(parent, current)
        for neighbour, action in transition(current, (x, y)): 
            if neighbour not in parent: 
                parent[neighbour] = (current, action)
                queue.append(neighbour)
    # No solution found in the reachable state space.
    return None
```

```python
# Helper functions, dividing the responsibilites

def goal_test(state, g): 
    return state[0] == g or state[1] == g

def transition(state, param): 
    # returns a list of possible successor states with the operator used to achieve that
    x, y = param
    a, b = state

    successors = []
    successors.append(((0, b), "Empty A")) # Empty a
    successors.append(((a, 0), "Empty B")) # Empty b
    successors.append(((x, b), "Fill A")) # Fill a
    successors.append(((a, y), "Fill B")) # Fill b
    t = min(y - b, a)
    successors.append(((a - t, b + t), "Pour A into B")) # Pour a into b
    t = min(x - a, b)
    successors.append(((a + t, b - t), "Pour B into A")) # Pour b into a

    return successors

def reconstruct_path(parent, state): 
    path = [(state, "GOAL REACHED")]
    current = state
    # Check while parent[current] is not None, as for (0,0) parent is None
    while parent[current] is not None: 
        path.append(parent[current])
        current = parent[current][0]
    path.reverse()
    return path
```

```python
def main(): 
    while True: 
        try: 
                
            x = int(input("Enter the capacity of first jug: "))
            y = int(input("Enter the capacity of second jug: "))
            g = int(input("Enter the goal amount (in litres): "))

            if x <= 0 or y <= 0:
                print("Jug capacities must be positive.")
                continue

            if g < 0:
                print("Target cannot be negative.")
                continue
                
            result = water_jug(x, y, g)
            if result is not None: 
                count = 1
                for element in result: 
                    print(f"Step {count}.", element[0], "->", element[1])
                    count += 1
            else: 
                print("The given condition is not solvable!")

        except:
            print("Enter integer values.")
            continue
                
        choice = input("Do you want to continue? (Y/n): ").strip().lower()
        if choice != "y": 
            break
```

```python
main()
```
