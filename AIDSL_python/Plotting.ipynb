```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 180, 170, 210, 250]
df = pd.read_csv('StudentsPerformance.csv')
```

```python
plt.figure(figsize=(7,6))
plt.plot(months, sales, marker='o')

plt.title('Monthly sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,6))

plt.bar(months, sales)

plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(sales, labels=months, autopct='%1.1f%%')

plt.title('Monthly Sales')

plt.show()
```

```python
dates = pd.date_range(start="2025-01-01", periods=10)

prices = [1200, 1225, 1240, 1230, 1265,
          1280, 1270, 1295, 1310, 1325]

ts = pd.DataFrame({
    "Date": dates,
    "Price": prices
})
plt.figure(figsize=(8, 4))

plt.plot(ts["Date"], ts["Price"], marker="o")

plt.title("Commodity Price Trend")
plt.xlabel("Date")
plt.ylabel("Price")

plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
ts["Moving_Average"] = ts["Price"].rolling(window=3).mean()

print(ts)
```

```python
plt.figure(figsize=(8, 4))

plt.plot(ts["Date"], ts["Price"],
         marker="o", label="Original Price")

plt.plot(ts["Date"], ts["Moving_Average"],
         marker="s", label="3-Day Moving Average")

plt.title("Price Trend with Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```
