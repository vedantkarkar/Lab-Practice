```python
import random
```

```python
def random_res_n_queens(N, max_restarts=100):
    for restarts in range(max_restarts): 
        state = [i for i in range(N)]
        random.shuffle(state)

        result = n_queens(state)

        if conflicts(result) == 0: 
            return (result, restarts)

    return (None, max_restarts)

def n_queens(state): 
    current = state.copy()
    while True:
        if conflicts(current) == 0:
            return current

        neighbours = gen_neighbours(current)
        
        min_conflict = float('inf')
        best = None
        for neighbour in neighbours: 
            if conflicts(neighbour) < min_conflict: 
                min_conflict = conflicts(neighbour)
                best = neighbour

        if min_conflict < conflicts(current): 
            current = best
        else:
            break
    return current
```

```python
def conflicts(state):
    n = len(state)
    conflict = 0
    for i in range(n): 
        for j in range(i+1, n): 
            if abs(i - j) == abs(state[i] - state[j]): 
                conflict += 1
    return conflict

def gen_neighbours(state): 
    n = len(state)
    neighbours = []
    for i in range(n): 
        for j in range(i+1, n): 
            neighbour = state.copy()
            neighbour[i], neighbour[j] = neighbour[j], neighbour[i]
            neighbours.append(neighbour)
    return neighbours

def draw_board(state): 
    n = len(state)
    for i in range(n): 
        for j in range(n): 
            if j == state[i]: 
                print(" Q ", sep="", end="")
            else: 
                print(" . ", sep="", end="")
        print()
```

```python
def main(): 
    while True: 
        try: 
            n = int(input("Enter the value of N: "))
            if n <= 0: 
                print("Please enter a positive integer")
                continue
            
            result, restarts = random_res_n_queens(n)

            if result is None: 
                print(f"Took \033[31m{restarts}\033[0m restarts.")
                print("Failed to find a solution")
            else:
                draw_board(result)
                print(f"Took \033[32m{restarts}\033[0m restarts.")
        except ValueError: 
            print("Enter a valid integer")
            continue

        choice = input("Do you want to continue? (Y/n): ").strip().lower()
        if choice != "y":
            break
```

```python
main()
```
