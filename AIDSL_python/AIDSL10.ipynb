```python

import numpy as np
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler

# 1. Load Dataset
df = pd.read_csv("titanic_data.csv")

# 2. Explore Dataset
print("\n--- Dataset Information ---")
print(df.head())
df.info()
print(df.describe())

# 3. Check Missing Values
print("\n--- Missing Values ---")
print(df.isnull().sum())

# 4. Handle Missing Numerical Values
df["age"] = df["age"].fillna(df["age"].mean())

# 5. Handle Missing Categorical Values
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

df["embark_town"] = df["embark_town"].fillna(
    df["embark_town"].mode()[0]
)

# Drop deck column (too many missing values)
df = df.drop("deck", axis=1)

print("\n--- Missing Values After Handling ---")
print(df.isnull().sum())

# 6. Remove Duplicate Records
print("\nDuplicate Records:", df.duplicated().sum())
df = df.drop_duplicates()

# 7. Label Encoding
encoder = LabelEncoder()
df["sex_encoded"] = encoder.fit_transform(df["sex"])

print("\n--- Label Encoding ---")
print(df[["sex", "sex_encoded"]].head())

# 8. One-Hot Encoding
df = pd.get_dummies(df, columns=["embarked"], dtype=int)

print("\n--- One-Hot Encoding ---")
print(df.head())

# 9. Standardization
scaler = StandardScaler()
df["age_standardized"] = scaler.fit_transform(df[["age"]])

# 10. Normalization
normalizer = MinMaxScaler()
df["fare_normalized"] = normalizer.fit_transform(df[["fare"]])

# 11. Feature Engineering
df["Family_Size"] = df["sibsp"] + df["parch"] + 1

df["Is_Alone"] = df["Family_Size"].apply(
    lambda x: 1 if x == 1 else 0
)

# 12. Display Processed Dataset
print("\n--- Processed Dataset ---")
print(df.head())

# 13. Save Processed Dataset
df.to_csv("Processed_Titanic.csv", index=False)

print("\nProcessed Dataset Saved Successfully!")
```
