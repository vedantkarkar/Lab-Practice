```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import norm
```

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
bins = np.linspace(0, 100, 11)
labels = [str(i) + "-" + str(i+10) for i in range(0,100,10)]

freq = pd.cut(df['math score'], bins=bins, labels=labels)
```

```python
freq_dist = freq.value_counts().sort_index()
```

```python
freq_dist
```

```python
marks = df['math score']

mean = marks.mean()
median = marks.median()
mode = marks.mode()[0]

var = marks.var()
std = marks.std()

print("Mean: ", mean)
print("Median: ", median)
print("Mode: ", mode)
print("Variance: ", var)
print("Standard deviation: ", std)
```

```python
plt.figure(figsize=(6, 6))

sns.histplot(data=df, x='math score', bins=10, color='skyblue')

plt.title("Histogram of math scores")
plt.xlabel("Score")
plt.ylabel("Number of students")

plt.grid(axis='y', linestyle="--", alpha=0.7)


plt.show()
```

```python
marks =df['math score']
x = np.linspace(marks.min(), marks.max(), 100)
y = norm.pdf(x, loc=marks.mean(), scale=marks.std())

plt.figure(figsize=(8, 8))

sns.histplot(data=df, x='math score', bins=25, color='lightgreen', label='Histogram', stat='density')
plt.plot(x, y, label='Normal Curve', color='red')

plt.title("Histogram and Normal Curve of math scores")
plt.xlabel("Score")
plt.ylabel("Density")
plt.legend()
plt.grid(True)

plt.show()
```

```python

# 1. Define your data series
data = [186, 176, 158, 180, 186, 168, 168, 164, 178, 170, 189, 195, 172,
        187, 180, 186, 185, 168, 179, 178, 183, 179, 170, 175, 186, 159,
        161, 178, 175, 185, 175, 162, 173, 172, 177, 175, 172, 177, 180]

# 2. Calculate mean and standard deviation
mean = np.mean(data)
std = np.std(data)

# 3. Generate x-values for the curve
x = np.linspace(min(data), max(data), 100)

# 4. Calculate y-values using the normal PDF
y = norm.pdf(x, mean, std)

# 5. Plot the curve
plt.figure(figsize=(10, 6))
plt.plot(x, y, color='blue', linewidth=2, label=f'Normal Curve (μ={mean:.2f}, σ={std:.2f})')
plt.title('Normal Distribution Curve')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()   
```
