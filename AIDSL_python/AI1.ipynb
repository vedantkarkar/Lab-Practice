```python
from collections import deque 
```

```python
def bfs(graph, start): 
    queue = deque([])
    queue.append(start)
    visited = set()
    visited.add(start)
    traversal = []
    while queue:
        node = queue.popleft()
        traversal.append(node)
        for neighbour in graph[node]:
            if neighbour not in visited:
                queue.append(neighbour)
                visited.add(neighbour)
    return traversal
```

```python
def dfs(graph, start): 
    stack = []
    stack.append(start)
    visited = set()
    visited.add(start)
    traversal = []
    while stack:
        node = stack.pop()
        traversal.append(node)
        for neighbour in reversed(graph[node]):
            if neighbour not in visited:
                stack.append(neighbour)
                visited.add(neighbour)
    return traversal
```

```python
def take_graph(): 
    # Takes graph as a adjacency list format
    graph = {}
    vertices = list()
    v = int(input("Input the number of vertices: "))
    for _ in range(v): 
        node = input("Enter name of one edge: ")
        vertices.append(node)
    for node in vertices: 
        print("Enter all the outlinks of '", node, "' seperated by <space>: ", end="")
        line = input().split()
        neighbours = []
        for element in line:
            if element in vertices:
                neighbours.append(element)
        graph[node] = neighbours
    return graph
```

```python
def take_undirected_graph():
    graph = {}
    n = int(input("Enter the number of vertices in the graph: "))
    while True:
        v = input("Enter the names of vertices (all in one line): ").split()
        if len(v) == n:
            break
        print("Please try again.")
    for vertex in v:
        graph[vertex] = []
    e = int(input("Enter the number of edges in the graph: "))
    count = 0
    while count < e:
        edge = input("Enter the two vertices of an edge (seperated by space, eg. A B): ").split()
        if len(edge) == 2 and edge[0] in v and edge[1] in v:
            graph[edge[0]].append(edge[1])
            graph[edge[1]].append(edge[0])
            count += 1
        else:
            print("Enter a valid edge connection")
    return graph
```

```python
def main(): 
    while True:
        graph = take_undirected_graph()
        print("From the following vertices, choose your start: ")
        for node in graph.keys(): 
            print(node)
        while True:
            start = input()
            if start in graph.keys():
                break
            print("Enter a valid vertex.")
        
        print("Traversing graph through DFS: ")
        print(dfs(graph, start))
        print("Traversing through BFS: ")
        print(bfs(graph, start))

        print("Do you want to continue? Y/n")
        choice = input()
        if choice != "Y" and choice != "y":
            print("Exiting program...")
            break
```
