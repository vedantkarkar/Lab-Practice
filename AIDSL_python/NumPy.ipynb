## Implementation test
Rows represent students; columns represent subjects.

Implement the following independently:
1. Print the shape, dimensions, and total number of elements.
2. Extract the marks of the third student.
3. Extract the marks of the second subject for all students.
4. Calculate the average marks of each student.
5. Calculate the average marks of each subject.
6. Add 5 bonus marks to every subject using broadcasting.
7. Calculate the overall average after adding the bonus.
8. Extract the first three students into a separate, independent array. Modify that array and verify that the original marks remains unchanged.

```python

# Implementation test

marks = np.array([
    [78, 85, 92, 67],
    [88, 76, 81, 90],
    [65, 72, 70, 80],
    [95, 91, 89, 94],
    [72, 68, 75, 79]
])

print("Shape: ", marks.shape)
print("Dimensions: ", marks.ndim)
print("Total elements: ", marks.size)

print("Marks of the third student: \n", marks[2, :])
print("Marks of the second subject: \n", marks[:, 1])
print("Average marks of each student: \n", np.average(marks, axis=0))
print("Average marks of each subject: \n", np.average(marks, axis=1))
print("Updated marks of the student after adding bonus: \n", marks + 5)
print("Overall average (After adding bonus): ", np.average(marks + 5))

copy_students = marks[:3, :].copy()
copy_students = copy_students - 15
print("Provisional marks of first 3 students: \n", copy_students)

print("Final Marks of the student: \n", marks)
```
