```python
import heapq
```

```python
# A star algorithm to work on a 2d grid or maze
def a_star(grid): 
    start = find_S(grid)
    goal = find_G(grid)

    frontier = [(h(start, goal), 0, start)]
    parent = {
        start : (0, None)
    }
    while frontier: 
        current_f, current_g, current = heapq.heappop(frontier)
        if goal_test(current, goal): 
            return reconstruct_path(current, parent)
            
        for successor in successors(current, grid): 
            g = current_g + 1
            f = g + h(successor, goal)
            if successor not in parent or g < parent[successor][0]: 
                parent[successor] = (g, current)
                heapq.heappush(frontier, (f, g, successor))
    return None

def draw_path(path, grid): 
    for i in range(len(grid)): 
        for j in range(len(grid[0])):
            if (i, j) in path:
                print("\033[32m*\033[0m", sep="", end="")
            else:
                print(grid[i][j], sep="", end="")
        print()
```

```python
# A star depends on this function

def h(state, goal): 
    return abs(state[0] - goal[0]) + abs(state[1] - goal[1])

def goal_test(state, goal): 
    return state == goal

def reconstruct_path(state, parent): 
    path = []
    current = state
    while parent[current][1] is not None: 
        path.append(current)
        current = parent[current][1]
    path.reverse()
    path.pop()
    return path

def blocked(state, grid): 
    r, c = state
    row, col = len(grid), len(grid[0])
    if r < 0 or r >= row or c < 0 or c >= col or grid[r][c] == "#": 
        return True
    return False

def successors(state, grid): 
    r, c = state
    moves = [
        (r+1, c), 
        (r, c+1), 
        (r-1, c), 
        (r, c-1)
    ]
    valid_successors = []
    for move in moves: 
        if not blocked(move, grid): 
            valid_successors.append(move)
    return valid_successors
```

```python
# defining the problem-specific semantics of the grid: 
# Must have a starting point S for start and G for goal
def sample_grid(): 
    grid = [
        "###################################################",
        "#                                        #        #",
        "#          #                             #        #",
        "#                                        #        #",
        "#                                        #        #",
        "#                                        #        #",
        "#                                        #        #",
        "#                      #                 #        #",
        "#                  #                     #        #",
        "#                                        #        #",
        "#          #                             #      #G#",
        "#                            ####        #        #",
        "#                                #       #        #",
        "#                                #      S#        #",
        "#           #          #         #########        #",
        "#                                        #      ###",
        "#      #                                          #",
        "###################################################"
    ]
    return grid

# this function should return the coordinates of start
def find_S(grid):
    row, col = len(grid), len(grid[0])
    for r in range(row): 
        for c in range(col): 
            if grid[r][c] == "S": 
                return (r, c)

# this function should return the coordinates of end
def find_G(grid):
    row, col = len(grid), len(grid[0])
    for r in range(row): 
        for c in range(col): 
            if grid[r][c] == "G": 
                return (r, c)
```

```python
# Put your grid here
grid = sample_grid()
path = a_star(grid)
if path is None: 
    print("No path found")
else:
    draw_path(path, grid)
```
