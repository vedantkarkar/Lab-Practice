```python
def main(): 
    game_over = False
    winner = None
    board = init_board()
    while not game_over: 
        display_board(board)

        choiceX = player_move(board) # takes a valid input from user
        if choiceX == -1: 
            break
        play(board, 'X', choiceX) # plays a turn and prints the turn

        if check_winner(board) == "X": 
            game_over = True
            winner = "X"
            break
        
        if is_draw(board): 
            game_over = True
            winner = 'draw'
            break
            
        choiceO = best_move(board)
        play(board, 'O', choiceO) 
        
        if check_winner(board) == "O": 
            game_over = True
            winner = "O"
            break
            
        if is_draw(board): 
            game_over = True
            winner = 'draw'
            break

    print("Game Over")
    display_board(board)
    if winner is None: 
        print("Player aborted the match")
    if winner == "O":
        print("Congratulations! Player 'O' wins the game")
    elif winner == "X":
        print("Congratulations! Player 'X' wins the game")
    elif winner == 'draw': 
        print("Match draws!")
```

```python
def init_board(): 
    return [i for i in range(1, 10)]

def display_board(board): 
    print('-'*11)
    for i in range(len(board)):
        if i % 3 == 2: 
            print(f" {board[i]} \n", end="")
            print('-'*11)
        else:
            print(f" {board[i]} |", end="")

def player_move(board): 
    while True:
        try: 
            choice = int(input("Enter a number available on the board: (or -1 to quit)"))
            if choice == -1:
                return -1
            if 1 <= choice <= 9 and board[choice-1] == choice: 
                return choice-1
            else:
                print("Invalid Input! ")
        except ValueError: 
            print("Enter a integer between 1 to 9")

def play(board, player, choice): 
    if 0 <= choice <= 8 and board[choice] == choice + 1:
        board[choice] = player
        print(f"Player {player} played on {choice+1}")

def is_draw(board): 
    for i in range(len(board)): 
        if board[i] == i+1: 
            return False
    return True

def check_winner(board): 
    win_combinations = [
        (1, 2, 3), 
        (4, 5, 6), 
        (7, 8, 9), 
        (1, 4, 7), 
        (2, 5, 8), 
        (3, 6, 9), 
        (1, 5, 9), 
        (3, 5, 7)
    ]
    for x, y, z in win_combinations: 
        if board[x-1] == board[y-1] == board[z-1]: 
            return board[x-1]
    return None

def possible_moves(board): 
    moves = []
    for i in range(len(board)): 
        if board[i] == i+1: 
            moves.append(i)
    return moves
```

```python
def minimax(board, isMax, depth): 
    if check_winner(board) == "X": 
        return 10 - depth
    if check_winner(board) == "O":
        return -10 + depth

    if is_draw(board):
        return 0

    if isMax:
        best = float('-inf')
        for move in possible_moves(board): 
            board[move] = 'X'
            score = minimax(board, False, depth+1)
            board[move] = move+1
            best = max(score, best)
        return best
    else: 
        best = float('inf')
        for move in possible_moves(board): 
            board[move] = 'O'
            score = minimax(board, True, depth+1)
            board[move] = move+1
            best = min(score, best)
        return best

# Hardcoded best move for 'O' (MIN)
def best_move(board): 
    best_move = None
    best_score = float('inf')

    for move in possible_moves(board): 
        board[move] = 'O'
        score = minimax(board, True, 0)
        board[move] = move+1

        if score < best_score: 
            best_score = score
            best_move = move

    return best_move



# def minimax(board, isMax): 
#     if check_winner(board) == 'X':
#         return 1

#     if check_winner(board) == 'O': 
#         return -1
        
#     if is_draw(board): 
#         return 0

#     if isMax: 
#         best = float('-inf')
#         for move in possible_moves(board): 
#             board[move] = 'X'
#             evaluation = minimax(board, False)
#             board[move] = move+1
#             best = max(evaluation, best)
#         return best
#     else: 
#         best = float('inf')
#         for move in possible_moves(board): 
#             board[move] = 'O'
#             evaluation = minimax(board, True)
#             board[move] = move+1
#             best = min(evaluation, best)
#         return best

# # here 'O' is hardcoded for best_move
# def best_move(board): 
#     best = float('inf')
#     best_move = None
#     for move in possible_moves(board): 
#         board[move] = 'O'
#         evaluation = minimax(board, True)
#         board[move] = move+1
#         if evaluation < best: 
#             best = evaluation
#             best_move = move

#     return best_move
```
