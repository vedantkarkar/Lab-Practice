```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("StudentsPerformance.csv")
# sns.set_theme(style="whitegrid", palette="deep", context="notebook")
sns.set_palette('muted')

# ----------------------------------
# PART A: BASIC MATPLOTLIB PLOTS
# ----------------------------------

# Line Plot
plt.figure(figsize=(7, 4))

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 200]
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()


# Bar Plot
plt.figure(figsize=(7, 4))

plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()


# Pie Chart
plt.figure(figsize=(6, 6))

plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()


# ----------------------------------
# PART B: SEABORN PLOTS
# ----------------------------------

# Scatter Plot
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=df,
    x="math score",
    y="reading score"
)

plt.title("Math vs Reading Scores")
plt.show()


# Histogram
plt.figure(figsize=(7, 4))

sns.histplot(
    data=df,
    x="math score",
    bins=15,
    kde=True
)

plt.title("Distribution of Math Scores")
plt.show()


# Box Plot
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=df,
    y="math score"
)

plt.title("Math Score Box Plot")
plt.show()


# Bar Plot using Seaborn
plt.figure(figsize=(7, 4))

sns.barplot(
    data=df,
    x="gender",
    y="math score"
)

plt.title("Average Math Score by Gender")
plt.show()


# ----------------------------------
# PART C: CORRELATION
# ----------------------------------

correlation = df["math score"].corr(df["writing score"])

print("Correlation:", correlation)


# Scatter Plot
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=df,
    x="math score",
    y="writing score"
)

plt.title("Math vs Writing")
plt.show()


# Regression Plot
plt.figure(figsize=(7, 4))

sns.regplot(
    data=df,
    x="math score",
    y="writing score"
)

plt.title("Regression Plot")
plt.show()


# ----------------------------------
# PART D: CORRELATION HEATMAP
# ----------------------------------

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()


# ----------------------------------
# PART E: TIME SERIES
# ----------------------------------

dates = pd.date_range(
    start="2025-01-01",
    periods=10
)

prices = [1200, 1225, 1240, 1230, 1265,
          1280, 1270, 1295, 1310, 1325]

ts = pd.DataFrame({
    "Date": dates,
    "Price": prices
})

# Moving Average
ts["Moving_Average"] = ts["Price"].rolling(window=3).mean()

print(ts)


# Plot Original Price and Moving Average
plt.figure(figsize=(8, 4))

plt.plot(
    ts["Date"],
    ts["Price"],
    marker='o',
    label="Original Price"
)

plt.plot(
    ts["Date"],
    ts["Moving_Average"],
    marker='s',
    label="Moving Average"
)

plt.title("Price Trend with Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```
