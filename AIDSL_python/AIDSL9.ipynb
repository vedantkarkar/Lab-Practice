```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import norm

sns.set_theme(palette=['skyblue'])
# Load Dataset
df = pd.read_csv("StudentsPerformance.csv")

marks = df["math score"]


# ----------------------------------
# PART A: CENTRAL TENDENCY
# ----------------------------------

mean_value = marks.mean()
median_value = marks.median()
mode_value = marks.mode()[0]

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)


# ----------------------------------
# PART B: FREQUENCY DISTRIBUTION
# ----------------------------------

bins = [0, 20, 40, 60, 80, 101]

labels = ["0-20", "20-40", "40-60", "60-80", "80-100"]

categories = pd.cut(
    marks,
    bins=bins,
    labels=labels,
    right=False
)

frequency = categories.value_counts().sort_index()

print("\nFrequency Distribution:")
print(frequency)


# ----------------------------------
# PART C: HISTOGRAM
# ----------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="math score",
    bins=10,
)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.show()


# ----------------------------------
# PART D: VARIANCE AND STANDARD DEVIATION
# ----------------------------------

variance_value = marks.var()
std_value = marks.std()

print("Variance:", variance_value)
print("Standard Deviation:", std_value)


# ----------------------------------
# PART E: NORMAL CURVE
# ----------------------------------

x = np.linspace(
    marks.min(),
    marks.max(),
    100
)

y = norm.pdf(
    x,
    loc=mean_value,
    scale=std_value
)

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="math score",
    bins=10,
    stat="density",
    label="Histogram"
)

plt.plot(
    x,
    y,
    color="steelblue",
    label="Normal Curve"
)

plt.title("Histogram with Normal Curve")
plt.xlabel("Math Score")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```
