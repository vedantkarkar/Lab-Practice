```python
completed_mask = df['test preparation course'] == 'completed'
none_mask = df['test preparation course'] == 'none'
```

```python
print(df[completed_mask].describe())
print(df[none_mask].describe())
```

```python
score_columns = [
    'math score',
    'reading score',
    'writing score'
]

completed_stats = df[completed_mask][score_columns].mean()
none_stats = df[none_mask][score_columns].mean()

comparison = pd.DataFrame({
    'Completed': completed_stats,
    'None': none_stats
})

print(comparison)
```
